## Predict one example transaction

The input must contain the same numeric feature columns used during training.

```python
import joblib
import pandas as pd
bundle = joblib.load('outputs/fraud_model.joblib')
example = pd.DataFrame([{feature: 0.0 for feature in bundle['features']}])
prediction = bundle['model'].predict(example)[0]
print('FRAUD' if prediction == 1 else 'LEGITIMATE')
if hasattr(bundle['model'], 'predict_proba'):
    print('Estimated fraud score:', bundle['model'].predict_proba(example)[0, 1])
```
